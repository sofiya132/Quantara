```python
import pandas as pd

X_train = pd.read_csv("X_train.csv")
X_test = pd.read_csv("X_test.csv")
y_train = pd.read_csv("y_train.csv")
y_test = pd.read_csv("y_test.csv")

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)

print("\nX_train:")
display(X_train.head())

print("\ny_train:")
display(y_train.head())
```

```python
print("Training target classes:")
print(y_train["Category"].value_counts())

print("\nTest target classes:")
print(y_test["Category"].value_counts())
```

```python
from sklearn.preprocessing import LabelEncoder


y_train_series = y_train["Category"]
y_test_series = y_test["Category"]


label_encoder = LabelEncoder()

y_train_encoded = label_encoder.fit_transform(y_train_series)
y_test_encoded = label_encoder.transform(y_test_series)

# Show the mapping
print("Class mapping:")
for number, label in enumerate(label_encoder.classes_):
    print(number, "=", label)

print("\nEncoded training labels:")
print(y_train_encoded[:10])

print("\nEncoded test labels:")
print(y_test_encoded[:10])
```

```python
from sklearn.linear_model import LogisticRegression

# Create the Logistic Regression model
logistic_model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

# Train the model
logistic_model.fit(X_train, y_train_encoded)

print("Logistic Regression trained successfully!")
```

```python
# Make predictions
y_pred_logistic = logistic_model.predict(X_test)

# Get prediction probabilities
y_prob_logistic = logistic_model.predict_proba(X_test)

print("Predictions:")
print(y_pred_logistic[:10])

print("\nProbability shape:", y_prob_logistic.shape)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

accuracy = accuracy_score(y_test_encoded, y_pred_logistic)
precision = precision_score(y_test_encoded, y_pred_logistic, average="weighted", zero_division=0)
recall = recall_score(y_test_encoded, y_pred_logistic, average="weighted", zero_division=0)
f1 = f1_score(y_test_encoded, y_pred_logistic, average="weighted", zero_division=0)

print("Logistic Regression Results")
print("----------------------------")
print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1 Score :", f1)

print("\nClassification Report:")
print(classification_report(
    y_test_encoded,
    y_pred_logistic,
    target_names=label_encoder.classes_,
    zero_division=0
))

print("Confusion Matrix:")
print(confusion_matrix(y_test_encoded, y_pred_logistic))
```

```python
from sklearn.ensemble import RandomForestClassifier

# Create Random Forest model
random_forest_model = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42
)

# Train the model
random_forest_model.fit(X_train, y_train_encoded)

print("Random Forest trained successfully!")
```

```python
# Make Random Forest predictions
y_pred_rf = random_forest_model.predict(X_test)

# Get prediction probabilities
y_prob_rf = random_forest_model.predict_proba(X_test)

print("Random Forest predictions:")
print(y_pred_rf[:10])

print("\nProbability shape:", y_prob_rf.shape)
```

```python
import numpy as np
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

def evaluate_model(y_true, y_pred, y_prob, model_name):

    # Basic metrics
    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(
        y_true, y_pred,
        average="weighted",
        zero_division=0
    )
    recall = recall_score(
        y_true, y_pred,
        average="weighted",
        zero_division=0
    )
    f1 = f1_score(
        y_true, y_pred,
        average="weighted",
        zero_division=0
    )

    # Multiclass ROC-AUC
    roc_auc = roc_auc_score(
        y_true,
        y_prob,
        multi_class="ovr",
        average="weighted"
    )

    # Confusion matrix
    cm = confusion_matrix(y_true, y_pred)

    # Sensitivity and specificity for each class
    sensitivities = []
    specificities = []

    for i in range(len(cm)):
        tp = cm[i, i]
        fn = cm[i, :].sum() - tp
        fp = cm[:, i].sum() - tp
        tn = cm.sum() - (tp + fn + fp)

        sensitivity = tp / (tp + fn) if (tp + fn) > 0 else 0
        specificity = tn / (tn + fp) if (tn + fp) > 0 else 0

        sensitivities.append(sensitivity)
        specificities.append(specificity)

    # Macro average across classes
    sensitivity_macro = np.mean(sensitivities)
    specificity_macro = np.mean(specificities)

    return {
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "ROC-AUC": roc_auc,
        "Sensitivity": sensitivity_macro,
        "Specificity": specificity_macro
    }
```

```python
logistic_results = evaluate_model(
    y_test_encoded,
    y_pred_logistic,
    y_prob_logistic,
    "Logistic Regression"
)

rf_results = evaluate_model(
    y_test_encoded,
    y_pred_rf,
    y_prob_rf,
    "Random Forest"
)

xgb_results = evaluate_model(
    y_test_encoded,
    y_pred_xgb,
    y_prob_xgb,
    "XGBoost"
)

print(logistic_results)
print(rf_results)
print(xgb_results)
```

```python
xgb_param_grid = {
    "n_estimators": [100, 200, 300],
    "max_depth": [2, 3, 4, 5],
    "learning_rate": [0.03, 0.05, 0.1],
    "subsample": [0.8, 1.0],
    "colsample_bytree": [0.8, 1.0]
}

xgb_grid = GridSearchCV(
    XGBClassifier(
        objective="multi:softprob",
        num_class=5,
        eval_metric="mlogloss",
        random_state=42
    ),
    xgb_param_grid,
    cv=cv,
    scoring="f1_weighted",
    n_jobs=-1
)

xgb_grid.fit(X_train, y_train_encoded)

print("Best XGBoost parameters:")
print(xgb_grid.best_params_)

print("\nBest CV Weighted F1:")
print(xgb_grid.best_score_)
```

```python
# Get the best tuned models
best_logistic = logistic_grid.best_estimator_
best_rf = rf_grid.best_estimator_
best_xgb = xgb_grid.best_estimator_

# Make predictions
y_pred_logistic_tuned = best_logistic.predict(X_test)
y_pred_rf_tuned = best_rf.predict(X_test)
y_pred_xgb_tuned = best_xgb.predict(X_test)

# Get probabilities
y_prob_logistic_tuned = best_logistic.predict_proba(X_test)
y_prob_rf_tuned = best_rf.predict_proba(X_test)
y_prob_xgb_tuned = best_xgb.predict_proba(X_test)

print("All tuned models evaluated on test data successfully!")
```

```python
%%writefile classical_models.py

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier


def create_logistic_regression():
    return LogisticRegression(
        C=0.01,
        class_weight="balanced",
        solver="lbfgs",
        max_iter=2000,
        random_state=42
    )


def create_random_forest():
    return RandomForestClassifier(
        n_estimators=200,
        max_depth=5,
        min_samples_split=2,
        class_weight="balanced",
        random_state=42
    )


def create_xgboost():
    return XGBClassifier(
        n_estimators=300,
        max_depth=4,
        learning_rate=0.1,
        subsample=0.8,
        colsample_bytree=1.0,
        objective="multi:softprob",
        num_class=5,
        eval_metric="mlogloss",
        random_state=42
    )


def train_models(X_train, y_train):
    models = {
        "Logistic Regression": create_logistic_regression(),
        "Random Forest": create_random_forest(),
        "XGBoost": create_xgboost()
    }

    for model in models.values():
        model.fit(X_train, y_train)

    return models
```

```python
%%writefile evaluation.py

import numpy as np

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)


def evaluate_model(y_true, y_pred, y_prob, model_name):
    """
    Evaluate a multiclass classification model.

    Returns:
        Dictionary containing Accuracy, Precision, Recall,
        F1, ROC-AUC, Sensitivity and Specificity.
    """

    # Basic metrics
    accuracy = accuracy_score(y_true, y_pred)

    precision = precision_score(
        y_true,
        y_pred,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        average="weighted",
        zero_division=0
    )

    # Multiclass ROC-AUC
    roc_auc = roc_auc_score(
        y_true,
        y_prob,
        multi_class="ovr",
        average="weighted"
    )

    # Confusion matrix
    cm = confusion_matrix(y_true, y_pred)

    sensitivities = []
    specificities = []

    # Calculate sensitivity and specificity for each class
    for i in range(len(cm)):

        tp = cm[i, i]
        fn = cm[i, :].sum() - tp
        fp = cm[:, i].sum() - tp
        tn = cm.sum() - (tp + fn + fp)

        sensitivity = (
            tp / (tp + fn)
            if (tp + fn) > 0
            else 0
        )

        specificity = (
            tn / (tn + fp)
            if (tn + fp) > 0
            else 0
        )

        sensitivities.append(sensitivity)
        specificities.append(specificity)

    # Macro average across classes
    sensitivity_macro = np.mean(sensitivities)
    specificity_macro = np.mean(specificities)

    return {
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "ROC-AUC": roc_auc,
        "Sensitivity": sensitivity_macro,
        "Specificity": specificity_macro
    }
```
